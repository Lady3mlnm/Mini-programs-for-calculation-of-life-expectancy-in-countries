```python
# create code for placing info in Wikipedia
def create_table_simlified(df, lang='en'):

    def if_value(x, prec=2):
        return '—' if math.isnan(x) else \
               f"{x:0.{prec}f}"  if x>=0 else \
               f"−{-x:0.{prec}f}"

    def chval(x, prec=2):  # change_value
        return '"| —' if math.isnan(x) else \
               f'color:darkgreen;"| {x:0.{prec}f}' if x>0 else \
               f'color:crimson;"| −{-x:0.{prec}f}' if x<0 else \
               f'color:darkgray;"| {x:0.{prec}f}'
    
    def chval_bold(x, prec=2):  # change_value
        return '"| —' if math.isnan(x) else \
               f'color:darkgreen;"| \'\'\'{x:0.{prec}f}\'\'\'' if x>0 else \
               f'color:crimson;"| \'\'\'−{-x:0.{prec}f}\'\'\'' if x<0 else \
               f'color:darkgray;"| \'\'\'{x:0.{prec}f}\'\'\''

    if lang=='ru':
        file_header='wbg_stats_header_2024_ru-simlified.txt'
        ptn_1 = 'флагификация'
        ptn_2 = 'флаг'
        prettify_name = {
            'European Union': '[[Европейский союз]]',
            'Central Europe and the Baltics': '[[Центральная Европа]] и [[Прибалтика]]',
            'Europe & Central Asia': '[[Европа]] и [[Центральная Азия]]',
            'South Asia': '[[Южная Азия]]',
            'East Asia & Pacific': '[[Восточная Азия]] и [[Австралия и Океания|Океания]]',
            'Middle East, North Africa, Afghanistan & Pakistan': '[[Страны MENA|Ближний Восток и Северная Африка]]',
            'Africa Eastern and Southern': '[[Восточная Африка|Восточная]] и [[Южная Африка|Южная]] Африка',
            'Sub-Saharan Africa': '[[Чёрная Африка|Африка к югу от Сахары]]',
            'Africa Western and Central': '[[Западная Африка|Западная]] и [[Центральная Африка|Центральная]] Африка',
            'North America': '[[Северная Америка]]',
            'Latin America & Caribbean': '[[Латинская Америка]] и [[Антильские острова|Карибы]]',
            'Arab World': '[[Арабский мир]]',
            'World': 'Мир',
            'Caribbean small states': '[[Список стран Карибского бассейна|Малые страны Карибского бассейна]]',
            'Pacific island small states': 'Малые страны Тихого океана',
            'Small states': 'Малые страны',
            'Other small states': 'Другие малые страны'
        }
    else:
        file_header='wbg_stats_header_2024_en-simlified.txt'
        ptn_1 = 'flaglist'
        ptn_2 = 'flagicon'
        prettify_name = {
            'European Union': '[[European Union]]',
            'Central Europe and the Baltics': '[[Central Europe]] & the [[Baltic states|Baltics]]',
            'Europe & Central Asia': '[[Europe]] & [[Central Asia]]',
            'South Asia': '[[South Asia]]',
            'East Asia & Pacific': '[[Asia-Pacific|East Asia & Pacific]]',
            'Middle East, North Africa, Afghanistan & Pakistan': '[[MENA|Middle East & North Africa]]',
            'Africa Eastern and Southern': 'Africa [[East Africa|Eastern]] & [[Southern Africa|Southern]]',
            'Sub-Saharan Africa': '[[Sub-Saharan Africa]]',
            'Africa Western and Central': 'Africa [[West Africa|Western]] & [[Central Africa|Central]]',
            'North America': '[[North America]]',
            'Latin America & Caribbean': '[[Latin America and the Caribbean|Latin America & Caribbean]]',
            'Arab World': '[[Arab world]]',
            'World': 'World'
        }
        
    with open('design/' + file_header, mode='r', encoding="utf-8") as fh:
        st_header = fh.read()
        
    st_header = st_header.strip()
    st = ''
        
    for i in range(len(df)):
        ser = df.iloc[i]
        if ser.name in ['World',
                        'European Union', 'Central Europe and the Baltics',
                        'Europe & Central Asia',
                        'South Asia',
                        'East Asia & Pacific',
                        'Middle East, North Africa, Afghanistan & Pakistan',
                        'Africa Eastern and Southern', 'Sub-Saharan Africa', 'Africa Western and Central',
                        'North America', 'Latin America & Caribbean']:
            st += '\n' + '|-class=static-row-header\n' + \
                  f'|style="text-align:center;"| \'\'\'{prettify_name.get(ser.name, ser.name)}\'\'\' ' + \
                  f'||style="background:#e0ffd8;"| \'\'\'{if_value(ser.total, 2)}\'\'\' ' + \
                  f'||style="background:#eaf3ff;"| \'\'\'{if_value(ser.male, 2)}\'\'\' ' + \
                  f'||style="background:#fee7f6;"| \'\'\'{if_value(ser.female, 2)}\'\'\' ' + \
                  f'||style="background:#fff8dc;"| \'\'\'{if_value(ser.gender_gap, 2)}\'\'\' ' + \
                  f'||style="border-left-width:2px;padding-left:1em;"| \'\'\'{if_value(ser["2014_t"], 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;{chval_bold(ser["2014→2019"], 2)} ' + \
                  f'|| \'\'\'{if_value(ser["2019_t"], 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;{chval_bold(ser["2019→2020"], 2)} ' + \
                  f'|| \'\'\'{if_value(ser["2020_t"], 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;{chval_bold(ser["2020→2021"], 2)} ' + \
                  f'|| \'\'\'{if_value(ser["2021_t"], 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;{chval_bold(ser["2021→2022"], 2)} ' + \
                  f'|| \'\'\'{if_value(ser["2022_t"], 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;{chval_bold(ser["2022→2023"], 2)} ' + \
                  f'|| \'\'\'{if_value(ser["2023_t"], 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;{chval_bold(ser["2023→2024"], 2)} ' + \
                  f'||style="background:#e0ffd8;"| \'\'\'{if_value(ser.total, 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;border-left-width:2px;{chval_bold(ser["2019→2024"], 2)}' + \
                  (f' ||style="background:#fffae0;{chval_bold(ser["2014→2024"], 2)}' if lang=='ru' else '')
        elif ser.name in ['Caribbean small states',
                          'Pacific island small states',
                          'Small states',
                          'Other small states',
                          'Kosove']:
            st += '\n' + '|-\n' + \
                  f'|style="padding-left:35px;"|{prettify_name.get(ser.name, ser.name)} ' + \
                  f'||style="background:#e0ffd8;"| {if_value(ser.total, 2)} ' + \
                  f'||style="background:#eaf3ff;"| {if_value(ser.male, 2)} ' + \
                  f'||style="background:#fee7f6;"| {if_value(ser.female, 2)} ' + \
                  f'||style="background:#fff8dc;"| {if_value(ser.gender_gap, 2)} ' + \
                  f'||style="border-left-width:2px;padding-left:1em;"| {if_value(ser["2014_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2014→2019"], 2)} ' + \
                  f'|| {if_value(ser["2019_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2019→2020"], 2)} ' + \
                  f'|| {if_value(ser["2020_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2020→2021"], 2)} ' + \
                  f'|| {if_value(ser["2021_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2021→2022"], 2)} ' + \
                  f'|| {if_value(ser["2022_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2022→2023"], 2)} ' + \
                  f'|| {if_value(ser["2023_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2023→2024"], 2)} ' + \
                  f'||style="background:#e0ffd8;"| {if_value(ser.total, 2)} ' + \
                  f'||style="background:#fffae0;border-left-width:2px;{chval(ser["2019→2024"], 2)}' + \
                  (f' ||style="background:#fffae0;{chval(ser["2014→2024"], 2)}' if lang=='ru' else '')
        else:
            st += '\n' + '|-\n' + \
                  f'| {{{{{ptn_1}|{ser.name}}}}} ' + \
                  f'||style="background:#e0ffd8;"| {if_value(ser.total, 2)} ' + \
                  f'||style="background:#eaf3ff;"| {if_value(ser.male, 2)} ' + \
                  f'||style="background:#fee7f6;"| {if_value(ser.female, 2)} ' + \
                  f'||style="background:#fff8dc;"| {if_value(ser.gender_gap, 2)} ' + \
                  f'||style="border-left-width:2px; padding-left:1em;"| {if_value(ser["2014_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2014→2019"], 2)} ' + \
                  f'|| {if_value(ser["2019_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2019→2020"], 2)} ' + \
                  f'|| {if_value(ser["2020_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2020→2021"], 2)} ' + \
                  f'|| {if_value(ser["2021_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2021→2022"], 2)} ' + \
                  f'|| {if_value(ser["2022_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2022→2023"], 2)} ' + \
                  f'|| {if_value(ser["2023_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2023→2024"], 2)} ' + \
                  f'||style="background:#e0ffd8;"| {if_value(ser.total, 2)} ' + \
                  f'||style="background:#fffae0;border-left-width:2px;{chval(ser["2019→2024"], 2)}' + \
                  (f' ||style="background:#fffae0;{chval(ser["2014→2024"], 2)}' if lang=='ru' else '')
    st += '\n|}'
    
    st = st.replace('color:darkgray;"| 0.00 ', 'color:darkgray;"| 0.00 ')  \
           .replace('color:crimson;"| −0.00 ', 'color:darkgray;"| 0.00 ')  \
           .replace('color:darkgreen;"| 0.00 ', 'color:darkgray;"| 0.00 ')  \
           .replace('color:darkgray;"| \'\'\'0.00\'\'\' ', 'color:darkgray;"| \'\'\'0.00\'\'\' ')  \
           .replace('color:crimson;"| \'\'\'−0.00\'\'\' ', 'color:darkgray;"| \'\'\'0.00\'\'\' ')  \
           .replace('color:darkgreen;"| \'\'\'0.00\'\'\' ', 'color:darkgray;"| \'\'\'0.00\'\'\' ')
    
    # WARNING: be sure that inline styles do not contains numbers with comma
    if (lang == 'ru') | (lang == 'fr'):
        st = re.sub('(?<=\\d)\\.(?=\\d)', ',', st)  # replace . to comma, if this . is between two digits

    # gray color for missing values
    st = st.replace(';"| —', ';color:silver;"| —') \
           .replace('|| —', '||style="color:silver;"| —')

    st = st_header + st
    
    return st
```

```python
# create code for placing info in Wikipedia
def create_table_for_regions(df, lang='en'):

    def if_value(x, prec=2):
        return '—' if math.isnan(x) else \
               f"{x:0.{prec}f}"  if x>=0 else \
               f"−{-x:0.{prec}f}"

    def chval(x, prec=2):  # change_value
        return '"| —' if math.isnan(x) else \
               f'color:darkgreen;"| {x:0.{prec}f}' if x>0 else \
               f'color:crimson;"| −{-x:0.{prec}f}' if x<0 else \
               f'color:darkgray;"| {x:0.{prec}f}'
    
    def chval_bold(x, prec=2):  # change_value
        return '"| —' if math.isnan(x) else \
               f'color:darkgreen;"| \'\'\'{x:0.{prec}f}\'\'\'' if x>0 else \
               f'color:crimson;"| \'\'\'−{-x:0.{prec}f}\'\'\'' if x<0 else \
               f'color:darkgray;"| \'\'\'{x:0.{prec}f}\'\'\''

    if lang=='ru':
        file_header='wbg_stats_header_2024_ru-simlified.txt'
        ptn_1 = 'флагификация'
        ptn_2 = 'флаг'
        prettify_name = {
            'European Union': '[[Европейский союз]]',
            'Central Europe and the Baltics': '[[Центральная Европа]] и [[Прибалтика]]',
            'Europe & Central Asia': '[[Европа]] и [[Центральная Азия]]',
            'South Asia': '[[Южная Азия]]',
            'East Asia & Pacific': '[[Восточная Азия]] и [[Австралия и Океания|Океания]]',
            'Middle East, North Africa, Afghanistan & Pakistan': '[[Страны MENA|Ближний Восток и Северная Африка]]',
            'Africa Eastern and Southern': '[[Восточная Африка|Восточная]] и [[Южная Африка|Южная]] Африка',
            'Sub-Saharan Africa': '[[Чёрная Африка|Африка к югу от Сахары]]',
            'Africa Western and Central': '[[Западная Африка|Западная]] и [[Центральная Африка|Центральная]] Африка',
            'North America': '[[Северная Америка]]',
            'Latin America & Caribbean': '[[Латинская Америка]] и [[Антильские острова|Карибы]]',
            'Arab World': '[[Арабский мир]]',
            'World': 'Мир'
        }
    else:
        file_header='wbg_stats_header_2024_en-simlified.txt'
        ptn_1 = 'flaglist'
        ptn_2 = 'flagicon'
        prettify_name = {
            'European Union': '[[European Union]]',
            'Central Europe and the Baltics': '[[Central Europe]] & the [[Baltic states|Baltics]]',
            'Europe & Central Asia': '[[Europe]] & [[Central Asia]]',
            'South Asia': '[[South Asia]]',
            'East Asia & Pacific': '[[Asia-Pacific|East Asia & Pacific]]',
            'Middle East, North Africa, Afghanistan & Pakistan': '[[MENA|Middle East & North Africa]]',
            'Africa Eastern and Southern': 'Africa [[East Africa|Eastern]] & [[Southern Africa|Southern]]',
            'Sub-Saharan Africa': '[[Sub-Saharan Africa]]',
            'Africa Western and Central': 'Africa [[West Africa|Western]] & [[Central Africa|Central]]',
            'North America': '[[North America]]',
            'Latin America & Caribbean': '[[Latin America and the Caribbean|Latin America & Caribbean]]',
            'Arab World': '[[Arab world]]',
            'World': 'World'
        }
        
    with open('design/' + file_header, mode='r', encoding="utf-8") as fh:
        st_header = fh.read()
        
    st_header = st_header.strip()
    st = ''
        
    for i in range(len(df)):
        ser = df.iloc[i]
        if ser.name in ['World']:
            st += '\n' + '|-class=static-row-header\n' + \
                  f'|style="text-align:center;"| \'\'\'{prettify_name.get(ser.name, ser.name)}\'\'\' ' + \
                  f'||style="background:#e0ffd8;"| \'\'\'{if_value(ser.total, 2)}\'\'\' ' + \
                  f'||style="background:#eaf3ff;"| \'\'\'{if_value(ser.male, 2)}\'\'\' ' + \
                  f'||style="background:#fee7f6;"| \'\'\'{if_value(ser.female, 2)}\'\'\' ' + \
                  f'||style="background:#fff8dc;"| \'\'\'{if_value(ser.gender_gap, 2)}\'\'\' ' + \
                  f'||style="border-left-width:2px; padding-left:1em;"| \'\'\'{if_value(ser["2014_t"], 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;{chval_bold(ser["2014→2019"], 2)} ' + \
                  f'|| \'\'\'{if_value(ser["2019_t"], 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;{chval_bold(ser["2019→2020"], 2)} ' + \
                  f'|| \'\'\'{if_value(ser["2020_t"], 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;{chval_bold(ser["2020→2021"], 2)} ' + \
                  f'|| \'\'\'{if_value(ser["2021_t"], 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;{chval_bold(ser["2021→2022"], 2)} ' + \
                  f'|| \'\'\'{if_value(ser["2022_t"], 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;{chval_bold(ser["2022→2023"], 2)} ' + \
                  f'|| \'\'\'{if_value(ser["2023_t"], 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;{chval_bold(ser["2023→2024"], 2)} ' + \
                  f'||style="background:#e0ffd8;"| \'\'\'{if_value(ser.total, 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;border-left-width:2px;{chval_bold(ser["2019→2024"], 2)}' + \
                  (f' ||style="background:#fffae0;{chval_bold(ser["2014→2024"], 2)}' if lang=='ru' else '')
        else:
            st += '\n' + '|-\n' + \
                  f'| {prettify_name.get(ser.name, ser.name)} ' + \
                  f'||style="background:#e0ffd8;"| {if_value(ser.total, 2)} ' + \
                  f'||style="background:#eaf3ff;"| {if_value(ser.male, 2)} ' + \
                  f'||style="background:#fee7f6;"| {if_value(ser.female, 2)} ' + \
                  f'||style="background:#fff8dc;"| {if_value(ser.gender_gap, 2)} ' + \
                  f'||style="border-left-width:2px; padding-left:1em;"| {if_value(ser["2014_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2014→2019"], 2)} ' + \
                  f'|| {if_value(ser["2019_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2019→2020"], 2)} ' + \
                  f'|| {if_value(ser["2020_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2020→2021"], 2)} ' + \
                  f'|| {if_value(ser["2021_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2021→2022"], 2)} ' + \
                  f'|| {if_value(ser["2022_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2022→2023"], 2)} ' + \
                  f'|| {if_value(ser["2023_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2023→2024"], 2)} ' + \
                  f'||style="background:#e0ffd8;"| {if_value(ser.total, 2)} ' + \
                  f'||style="background:#fffae0;border-left-width:2px;{chval(ser["2019→2024"], 2)}' + \
                  (f' ||style="background:#fffae0;{chval(ser["2014→2024"], 2)}' if lang=='ru' else '')
    st += '\n|}'
    
    st = st.replace('color:darkgray;"| 0.00 ', 'color:darkgray;"| 0.00 ')  \
           .replace('color:crimson;"| −0.00 ', 'color:darkgray;"| 0.00 ')  \
           .replace('color:darkgreen;"| 0.00 ', 'color:darkgray;"| 0.00 ')  \
           .replace('color:darkgray;"| \'\'\'0.00\'\'\' ', 'color:darkgray;"| \'\'\'0.00\'\'\' ')  \
           .replace('color:crimson;"| \'\'\'−0.00\'\'\' ', 'color:darkgray;"| \'\'\'0.00\'\'\' ')  \
           .replace('color:darkgreen;"| \'\'\'0.00\'\'\' ', 'color:darkgray;"| \'\'\'0.00\'\'\' ')

    # WARNING: be sure that inline styles do not contains numbers with comma
    if (lang == 'ru') | (lang == 'fr'):
        st = re.sub('(?<=\\d)\\.(?=\\d)', ',', st)  # replace . to comma, if this . is between two digits
        
    st = st_header + st

    return st
```

```python
# create code for placing info in Wikipedia
def create_table_China(df, lang='en'):

    def if_value(x, prec=2):
        return '—' if math.isnan(x) else \
               f"{x:0.{prec}f}"  if x>=0 else \
               f"−{-x:0.{prec}f}"

    def chval(x, prec=2):  # change_value
        return '"| —' if math.isnan(x) else \
               f'color:darkgreen;"| {x:0.{prec}f}' if x>0 else \
               f'color:crimson;"| −{-x:0.{prec}f}' if x<0 else \
               f'color:darkgray;"| {x:0.{prec}f}'
    
    def chval_bold(x, prec=2):  # change_value
        return '"| —' if math.isnan(x) else \
               f'color:darkgreen;"| \'\'\'{x:0.{prec}f}\'\'\'' if x>0 else \
               f'color:crimson;"| \'\'\'−{-x:0.{prec}f}\'\'\'' if x<0 else \
               f'color:darkgray;"| \'\'\'{x:0.{prec}f}\'\'\''

    if lang=='ru':
        file_header='wbg_stats_header_2024_ru-simlified.txt'
        ptn_1 = 'флагификация'
        ptn_2 = 'флаг'
        prettify_name = {
            'China': 'Китай в среднем'
        }
    else:
        file_header='wbg_stats_header_2024_en-simlified.txt'
        ptn_1 = 'flaglist'
        ptn_2 = 'flagicon'
        prettify_name = {
             'China': 'China on average'
        }
        
    with open('design/' + file_header, mode='r', encoding="utf-8") as fh:
        st_header = fh.read()
        
    st_header = st_header.strip()
    st = ''
        
    for i in range(len(df)):
        ser = df.iloc[i]
        if ser.name in ['China']:
            st += '\n' + '|-class=static-row-header\n' + \
                  f'|style="text-align:center;"| \'\'\'{prettify_name.get(ser.name, ser.name)}\'\'\' ' + \
                  f'||style="background:#e0ffd8;"| \'\'\'{if_value(ser.total, 2)}\'\'\' ' + \
                  f'||style="background:#eaf3ff;"| \'\'\'{if_value(ser.male, 2)}\'\'\' ' + \
                  f'||style="background:#fee7f6;"| \'\'\'{if_value(ser.female, 2)}\'\'\' ' + \
                  f'||style="background:#fff8dc;"| \'\'\'{if_value(ser.gender_gap, 2)}\'\'\' ' + \
                  f'||style="border-left-width:2px;padding-left:1em;"| \'\'\'{if_value(ser["2014_t"], 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;{chval_bold(ser["2014→2019"], 2)} ' + \
                  f'|| \'\'\'{if_value(ser["2019_t"], 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;{chval_bold(ser["2019→2020"], 2)} ' + \
                  f'|| \'\'\'{if_value(ser["2020_t"], 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;{chval_bold(ser["2020→2021"], 2)} ' + \
                  f'|| \'\'\'{if_value(ser["2021_t"], 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;{chval_bold(ser["2021→2022"], 2)} ' + \
                  f'|| \'\'\'{if_value(ser["2022_t"], 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;{chval_bold(ser["2022→2023"], 2)} ' + \
                  f'|| \'\'\'{if_value(ser["2023_t"], 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;{chval_bold(ser["2023→2024"], 2)} ' + \
                  f'||style="background:#e0ffd8;"| \'\'\'{if_value(ser.total, 2)}\'\'\' ' + \
                  f'||style="background:#fffae0;border-left-width:2px;{chval_bold(ser["2019→2024"], 2)}' + \
                  (f' ||style="background:#fffae0;{chval_bold(ser["2014→2024"], 2)}' if lang=='ru' else '')
        else:
            st += '\n' + '|-\n' + \
                  f'| {{{{{ptn_1}|{ser.name}}}}} ' + \
                  f'||style="background:#e0ffd8;"| {if_value(ser.total, 2)} ' + \
                  f'||style="background:#eaf3ff;"| {if_value(ser.male, 2)} ' + \
                  f'||style="background:#fee7f6;"| {if_value(ser.female, 2)} ' + \
                  f'||style="background:#fff8dc;"| {if_value(ser.gender_gap, 2)} ' + \
                  f'||style="border-left-width:2px;padding-left:1em;"| {if_value(ser["2014_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2014→2019"], 2)} ' + \
                  f'|| {if_value(ser["2019_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2019→2020"], 2)} ' + \
                  f'|| {if_value(ser["2020_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2020→2021"], 2)} ' + \
                  f'|| {if_value(ser["2021_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2021→2022"], 2)} ' + \
                  f'|| {if_value(ser["2022_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2022→2023"], 2)} ' + \
                  f'|| {if_value(ser["2023_t"], 2)} ' + \
                  f'||style="background:#fffae0;{chval(ser["2023→2024"], 2)} ' + \
                  f'||style="background:#e0ffd8;"| {if_value(ser.total, 2)} ' + \
                  f'||style="background:#fffae0;border-left-width:2px;{chval(ser["2019→2024"], 2)} ' + \
                  (f' ||style="background:#fffae0;{chval(ser["2014→2024"], 2)}' if lang=='ru' else '')
    st += '\n|}'
    
    # WARNING: be sure that inline styles do not contains numbers with comma
    if (lang == 'ru') | (lang == 'fr'):
        st = re.sub('(?<=\\d)\\.(?=\\d)', ',', st)  # replace . to comma, if this . is between two digits
        
    st = st_header + st
    
    return st
```
