Data source: [2017–2019](https://www.ons.gov.uk/peoplepopulationandcommunity/healthandsocialcare/healthandlifeexpectancies/bulletins/lifeexpectancyforlocalareasoftheuk/between2001to2003and2017to2019), [2022–2024](https://www.ons.gov.uk/peoplepopulationandcommunity/healthandsocialcare/healthandlifeexpectancies/bulletins/lifeexpectancyforlocalareasoftheuk/between2001to2003and2022to2024)

```python
import pandas as pd
import math
import re

import sys
sys.path.append("..")
import mal_moduls_private.mal_total as mal
```

```python
df_2019 = pd.read_csv('data/LE in regions of England in 2017-19.csv', engine='python', skiprows=6, skipfooter=5, header=None, index_col=0, usecols=[0,1,6])

# manually patch DataFrame with values for the United Kingdom on average
if 'United Kingdom' not in df_2019.index:
    df_2019.loc['United Kingdom'] = [79.4, 83.1]

df_2019.rename(index={'Yorkshire and The Humber': 'Yorkshire and the Humber'}, inplace=True)  #  'East of England': 'East'

df_2019.index.name = ''
df_2019.columns = ['2017-19_m', '2017-19_f']
df_2019['2017-19_fΔm'] = (df_2019['2017-19_f'] - df_2019['2017-19_m']).round(1)
df_2019['2017-19_mean'] = ((df_2019['2017-19_f'] + df_2019['2017-19_m']) / 2).round(2)
df_2019
```

```python
df_2024 = pd.read_excel('data/LE in regions of England in 2022-24 -figure 1 data.xlsx', skiprows=6, index_col=0, usecols=[0,1,4])

# df_2024.rename(index={'East': 'East of England'}, inplace=True)

df_2024.index.name = ''
df_2024.columns = ['2022-24_m', '2022-24_f']
df_2024['2022-24_fΔm'] = (df_2024['2022-24_f'] - df_2024['2022-24_m']).round(1)
df_2024['2022-24_mean'] = ((df_2024['2022-24_f'] + df_2024['2022-24_m']) / 2).round(2)
df_2024
```

```python
df = pd.concat([df_2019, df_2024], axis='columns')

df.insert(loc=4, column='Δ_in_mean', value=(df['2022-24_mean']-df['2017-19_mean']).round(2))

df.sort_values(by=['2017-19_mean', '2017-19_m'], ascending=False, inplace=True)

df.fillna('')
```

```python
dd_replacement = {
    'England' : {'en': ('England on average', 'England'), 'ru': ('Англия в среднем', 'Англия')},
    'Northern Ireland' : {'en': ('Northern Ireland', 'Northern Ireland'), 'ru': ('Северная Ирландия', 'Северная Ирландия')},
    'Scotland' : {'en': ('Scotland', 'Scotland'), 'ru': ('Шотландия', 'Шотландия')},
    'Wales' : {'en': ('Wales', 'Wales'), 'ru': ('Уэльс', 'Уэльс')},
    'East of England' : {'en': ('East of England', 'East of England'), 'ru': ('Восточная Англия', 'Восточная Англия')},
    'East Midlands' : {'en': ('East Midlands', 'East Midlands'), 'ru': ('Восточный Ми́дленд', 'Восточный Мидленд')},
    'London' : {'en': ('London region', 'Greater London'), 'ru': ('Большой Ло́ндон', 'Большой Лондон')},
    'North East' : {'en': ('North East England', 'North East England'), 'ru': ('Северо-Восточная Англия', 'Северо-Восточная Англия')},
    'North West' : {'en': ('North West England', 'North West England'), 'ru': ('Северо-Западная Англия', 'Северо-Западная Англия')},
    'South East' : {'en': ('South East England', 'South East England'), 'ru': ('Юго-Восточная Англия', 'Юго-Восточная Англия')},
    'South West' : {'en': ('South West England', 'South West England'), 'ru': ('Юго-Западная Англия', 'Юго-Западная Англия')},
    'West Midlands' : {'en': ('West Midlands', 'West Midlands (region)'), 'ru': ('Западный Ми́дленд', 'Западный Мидленд')},
    'Yorkshire and the Humber' : {'en': ('Yorkshire and the Humber', 'Yorkshire and the Humber'), 'ru': ('Йо́ркшир и Ха́мбер', 'Йоркшир и Хамбер')},
    'United Kingdom' : {'en': ('United Kingdom', 'United Kingdom'), 'ru': ('Великобрита́ния ', 'Великобритания')}
}
```

```python
# create code for placing info in Wikipedia
def create_table(df, file_header, lang='ru'):

    def if_value(x, prec=1):
        return '—' if math.isnan(x) else \
               f"{x:0.{prec}f}"  if x>=0 else \
               f"−{-x:0.{prec}f}"                #"{x:0.{prec}f}".format(x, prec)
    
    def chval(x, prec=1, *, add_par=''):  # change_value
        return f'style="background:#fffae0;{add_par}"|—' if math.isnan(x) else \
               f'style="background:#fffae0;color:darkgreen;{add_par}"|{x:0.{prec}f}' if x>0 else \
               f'style="background:#fffae0;color:crimson;{add_par}"|−{-x:0.{prec}f}' if x<0 else \
               f'style="background:#fffae0;color:darkgray;{add_par}"|{x:0.{prec}f}'
    
    def chval_bold(x, prec=1, *, add_par=''):  # change_value
        return '—' if math.isnan(x) else \
               f'style="background:#fffae0;color:darkgreen;{add_par}"|\'\'\'{x:0.{prec}f}\'\'\'' if x>0 else \
               f'style="background:#fffae0;color:crimson;{add_par}"|\'\'\'−{-x:0.{prec}f}\'\'\'' if x<0 else \
               f'style="background:#fffae0;color:darkgray;{add_par}"|\'\'\'{x:0.{prec}f}\'\'\''
    
    with open('design/' + file_header, mode='r', encoding="utf-8") as fh:
        table_header = fh.read()

    st = ''
    for i in range(len(df)):
        ser = df.iloc[i]
        name_link = dd_replacement[ser.name][lang][1]
        name_visible = dd_replacement[ser.name][lang][0]
        name_inserted = name_link if name_link == name_visible else f"{name_link}|{name_visible}"

        if ser.name in ['England', 'Northern Ireland', 'Scotland', 'Wales', 'United Kingdom']:
             st += '\n' + ('|-class=static-row-header\n' if ser.name in ['England', 'United Kingdom'] else '|-\n') + \
                  f'| \'\'\'[[{name_inserted}]]\'\'\' ' + \
                  f'||style="background:#eaf3ff;"| \'\'\'{if_value(ser["2017-19_m"], prec=1)}\'\'\' ' + \
                  f'||style="background:#fee7f6;"| \'\'\'{if_value(ser["2017-19_f"], prec=1)}\'\'\' ' + \
                  f'||style="background:#fff8dc;"| \'\'\'{if_value(ser["2017-19_fΔm"], prec=1)}\'\'\' ' + \
                  f'|| \'\'\'{if_value(ser["2017-19_mean"], prec=2)}\'\'\' ' + \
                  f'||style="border-left-width:2px;padding-right:3ex;"| \'\'\'{if_value(ser["Δ_in_mean"], prec=2)}\'\'\' ' + \
                  f'||style="background:#eaf3ff;border-left-width:2px;"| \'\'\'{if_value(ser["2022-24_m"], prec=1)}\'\'\' ' + \
                  f'||style="background:#fee7f6;"| \'\'\'{if_value(ser["2022-24_f"], prec=1)}\'\'\' ' + \
                  f'||style="background:#fff8dc;"| \'\'\'{if_value(ser["2022-24_fΔm"], prec=1)}\'\'\' ' + \
                  f'|| \'\'\'{if_value(ser["2022-24_mean"], prec=2)}\'\'\''
        else:
            st += '\n' + '|-\n' + \
                  f'| [[{name_inserted}]] ' + \
                  f'||style="background:#eaf3ff;"| {if_value(ser["2017-19_m"], prec=1)} ' + \
                  f'||style="background:#fee7f6;"| {if_value(ser["2017-19_f"], prec=1)} ' + \
                  f'||style="background:#fff8dc;"| {if_value(ser["2017-19_fΔm"], prec=1)} ' + \
                  f'|| {if_value(ser["2017-19_mean"], prec=2)} ' + \
                  f'||style="border-left-width:2px;padding-right:3ex;"| {if_value(ser["Δ_in_mean"], prec=2)} ' + \
                  f'||style="background:#eaf3ff;border-left-width:2px;"| {if_value(ser["2022-24_m"], prec=1)} ' + \
                  f'||style="background:#fee7f6;"| {if_value(ser["2022-24_f"], prec=1)} ' + \
                  f'||style="background:#fff8dc;"| {if_value(ser["2022-24_fΔm"], prec=1)} ' + \
                  f'|| {if_value(ser["2022-24_mean"], prec=2)}'
        
    if lang == 'ru':
        st = re.sub('(?<=\\d)\\.(?=\\d)', ',', st)  # replace . to comma, if this . is between two digits

    st = table_header + st + '\n|}'
    
    # gray color for missing values
    st = st.replace(';"| —', ';color:silver;"| —')

    return st



table_code = create_table(df, file_header='UK_header-2024-en.txt', lang='en')
with open('output/Table code for British regions -en.txt', 'w', encoding="utf-8") as fh:
    fh.write(table_code)

table_code = create_table(df, file_header='UK_header-2024-ru.txt', lang='ru')
with open('output/Table code for British regions -ru.txt', 'w', encoding="utf-8") as fh:
    fh.write(table_code)
```
