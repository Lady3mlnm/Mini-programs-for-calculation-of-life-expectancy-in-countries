```python
def load_data_from_csv(file_name, selected_indicator, selected_year=YEAR_SELECTED, prefix_for_cols=''):
    df = pd.read_csv(f"data/{file_name}",
                     usecols=['Indicator', 'Location', 'Period', 'Dim1', 'FactValueNumeric'])

    df.rename(columns={'Location': 'country',
                       'Period': 'year',
                       'Dim1': 'sex',
                       'FactValueNumeric': 'value'}, inplace=True)

    
    df = df[(df.Indicator == selected_indicator) & (df.year == selected_year)] \
              .drop(columns=['Indicator', 'year'])

    df = df[~df['country'].isin(dd_countries_ignore)]
    df['country'] = df['country'].replace(dd_countries_renaming)
    df = df.pivot(index='country', columns='sex', values='value')

    df = df[['Both sexes', 'Male', 'Female']]                       # just reorder
    df.rename(columns = {'Both sexes': f"{prefix_for_cols}o",
                         'Male': f"{prefix_for_cols}m",
                         'Female': f"{prefix_for_cols}f"}, inplace=True)
    df.index.name = ''
    df.columns.name = ''
    df[f"{prefix_for_cols}fΔm"] = (df[f"{prefix_for_cols}f"] - df[f"{prefix_for_cols}m"]).round(2)
    return df

df_le = load_data_from_csv(file_name="Life expectancy at birth -countries.csv",
                           selected_indicator="Life expectancy at birth (years)",
                           prefix_for_cols="LE_")
df_le
```
