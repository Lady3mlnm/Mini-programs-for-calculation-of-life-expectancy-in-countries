```python
import pandas as pd
import numpy as np
import math
import re
from collections import namedtuple
```

```python
df1 = pd.read_csv('data/NFHS_2015-16.csv', sep='\t', index_col='state')
df1.index.name = ''
print(len(df1))

df1.head(3).fillna('')
```

```python
df2 = pd.read_csv('data/NFHS_2019-21.csv', sep='\t', index_col='state')
df2.index.name = ''
print(len(df2))

df2.head(3).fillna('')
```

```python
# just for interest
df_SRS = pd.concat([df1.iloc[:, 3:], df2.iloc[:, 3:]], axis='columns')

df_SRS.rename(columns = {'SRS-2015_total' : '2015_total',
                         'SRS-2015_male'  : '2015_male',
                         'SRS-2015_female': '2015_female',
                         'SRS-2020_total' : '2020_total',
                         'SRS-2020_male'  : '2020_male',
                         'SRS-2020_female': '2020_female'                         
                         }, inplace=True)
print(len(df_SRS))

df_SRS.dropna(how='all', inplace=True)
print(len(df_SRS))

df_SRS.sort_values(by=['2015_total', '2015_male'], ascending=False, inplace=True)
df_SRS = pd.concat([df_SRS.loc[['India']], df_SRS.drop('India')])

df_SRS.insert(loc=3, column='2015_fΔm' ,  value=(df_SRS['2015_female']-df_SRS['2015_male']).round(1))
df_SRS.insert(loc=4, column='Δ_periods',  value=(df_SRS['2020_total']-df_SRS['2015_total']).round(1))
df_SRS.insert(loc=8, column='2020_fΔm' ,  value=(df_SRS['2020_female']-df_SRS['2020_male']).round(1))

df_SRS.fillna('')
```

```python
df_NFHS = pd.concat([df1.iloc[:, :3], df2.iloc[:, :3]], axis='columns')

df_NFHS.rename(columns = {'NFHS-2015-16_total' : '2015-16_total',
                          'NFHS-2015-16_male'  : '2015-16_male',
                          'NFHS-2015-16_female': '2015-16_female',
                          'NFHS-2019-21_total' : '2019-21_total',
                          'NFHS-2019-21_male'  : '2019-21_male',
                          'NFHS-2019-21_female': '2019-21_female'                          
                         }, inplace=True)

print(len(df_NFHS))

df_NFHS.sort_values(by=['2015-16_total', '2015-16_male'], ascending=False, inplace=True)
df_NFHS = pd.concat([df_NFHS.loc[['India']], df_NFHS.drop('India')])

df_NFHS.insert(loc=3, column='2015-16_fΔm',  value=(df_NFHS['2015-16_female']-df_NFHS['2015-16_male']).round(1))
df_NFHS.insert(loc=4, column='Δ_periods'  ,  value=(df_NFHS['2019-21_total']-df_NFHS['2015-16_total']).round(1))
df_NFHS.insert(loc=8, column='2019-21_fΔm',  value=(df_NFHS['2019-21_female']-df_NFHS['2019-21_male']).round(1))

df_NFHS.fillna('')
```

```python
# handle special case
if all(df_NFHS.loc['Dadra and Nagar haveli', '2019-21_total':'2019-21_fΔm'] == df_NFHS.loc['Daman and diu', '2019-21_total':'2019-21_fΔm']):
    new_record = df_NFHS.loc['Dadra and Nagar haveli', '2019-21_total':'2019-21_fΔm']
    new_record.name = 'Dadra and Nagar Haveli and Daman and Diu'
    # df_NFHS = df_NFHS.append(new_record)
    df_NFHS = pd.concat([df_NFHS, new_record.to_frame().T], axis='index')
    df_NFHS.loc['Dadra and Nagar haveli', 'Δ_periods':'2019-21_fΔm'] = [np.NaN] * 5
    df_NFHS.loc['Daman and diu', 'Δ_periods':'2019-21_fΔm'] = [np.NaN] * 5

df_NFHS.fillna('')
```

```python
# Some inner statistics for interest:
def min_and_max(df, region_center=''):
    print(f"Number of values: {len(df)}")
    
    def find_3_largest(df):
        t_df = df.nlargest(3)
        t_df = t_df.reset_index()
        return t_df.apply(lambda v: f"{v.iloc[1]:.1f} – {v.iloc[0]}", axis=1)

    def find_3_smallest(df):
        t_df = df.nsmallest(3).iloc[::-1]
        t_df = t_df.reset_index()
        return t_df.apply(lambda v: f"{v.iloc[1]:.1f} – {v.iloc[0]}", axis=1)
    
    if region_center:
        ser_center = df.loc[region_center].map(lambda v: f" — {v} —")
        final_df = pd.concat([df.apply(find_3_largest), ser_center.to_frame().T, df.apply(find_3_smallest)])
        final_df.index = ['max', 'max_2', 'max_3', region_center, 'min_3', 'min_2', 'min']
    else:
        final_df = pd.concat([df.apply(find_3_largest), df.apply(find_3_smallest)])
        final_df.index = ['max', 'max_2', 'max_3', 'min_3', 'min_2', 'min']
    return final_df.style.set_properties(**{'text-align': 'left'})

min_and_max(df_SRS, region_center="India")
```

```python
# create code for placing info in Wikipedia
def create_table(df, file_header, lang='ru'):

    def if_value(x, prec=1):
        return '—' if math.isnan(x) else \
               f"{x:0.{prec}f}"  if x>=0 else \
               f"−{-x:0.{prec}f}"                #"{x:0.{prec}f}".format(x, prec)

    def chval(x, prec=1, *, add_par=''):  # change_value
        return f'style="padding-right:2ex;{add_par}"|—' if math.isnan(x) else \
               f'style="padding-right:2ex;color:darkgreen;{add_par}"|{x:0.{prec}f}' if x>0 else \
               f'style="padding-right:2ex;color:crimson;{add_par}"|−{-x:0.{prec}f}' if x<0 else \
               f'style="padding-right:2ex;color:darkgray;{add_par}"|{x:0.{prec}f}'
    
    def chval_bold(x, prec=1, *, add_par=''):  # change_value
        return '—' if math.isnan(x) else \
               f'style="padding-right:2ex;color:darkgreen;{add_par}"|\'\'\'{x:0.{prec}f}\'\'\'' if x>0 else \
               f'style="padding-right:2ex;color:crimson;{add_par}"|\'\'\'−{-x:0.{prec}f}\'\'\'' if x<0 else \
               f'style="padding-right:2ex;color:darkgray;{add_par}"|\'\'\'{x:0.{prec}f}\'\'\''
    
    with open('design/' + file_header, mode='r', encoding="utf-8") as fh:
        table_header = fh.read()

    st = ''
    for i in range(len(df)):
        ser = df.iloc[i]    
        if ser.name == 'India':
            st += '\n' + '|-class=static-row-header\n' + \
                  f'|style="text-align:left;"|\'\'\'{dd_replacement[ser.name][lang][0]}\'\'\' ' + \
                  f'|| style="text-align:center; background:#e0ffd8;"|\'\'\'{if_value(ser["2015-16_total"])}\'\'\' ' + \
                  f'|| style="text-align:center;background:#eaf3ff;"|\'\'\'{if_value(ser["2015-16_male"])}\'\'\' ' + \
                  f'|| style="text-align:center;background:#fee7f6;"|\'\'\'{if_value(ser["2015-16_female"])}\'\'\' ' + \
                  f'|| style="padding-right:2ex;"|\'\'\'{if_value(ser["2015-16_fΔm"])}\'\'\' ' + \
                  f'|| {chval_bold(ser["Δ_periods"], add_par="border-left-width:2px;")} ' + \
                  f'|| style="text-align:center;background:#e0ffd8;border-left-width:2px;"|\'\'\'{if_value(ser["2019-21_total"])}\'\'\' ' + \
                  f'|| style="text-align:center;background:#eaf3ff;"|\'\'\'{if_value(ser["2019-21_male"])}\'\'\' ' + \
                  f'|| style="text-align:center;background:#fee7f6;"|\'\'\'{if_value(ser["2019-21_female"])}\'\'\' ' + \
                  f'|| style="padding-right:2ex;"|\'\'\'{if_value(ser["2019-21_fΔm"])}\'\'\''
        else:
            name_link = dd_replacement[ser.name][lang][1]
            name_visible = dd_replacement[ser.name][lang][0]
            name_inserted = name_link if name_link == name_visible else f"{name_link}|{name_visible}"
            st += '\n' + '|-\n' + \
                  f'|style="text-align:left;"|[[{name_inserted}]] ' + \
                  f'|| style="text-align:center; background:#e0ffd8;"|\'\'\'{if_value(ser["2015-16_total"])}\'\'\' ' + \
                  f'|| style="text-align:center;background:#eaf3ff;"|{if_value(ser["2015-16_male"])} ' + \
                  f'|| style="text-align:center;background:#fee7f6;"|{if_value(ser["2015-16_female"])} ' + \
                  f'|| style="padding-right:2ex;"|{if_value(ser["2015-16_fΔm"])} ' + \
                  f'|| {chval(ser["Δ_periods"], add_par="border-left-width:2px;")} ' + \
                  f'|| style="text-align:center;background:#e0ffd8;border-left-width:2px;"|\'\'\'{if_value(ser["2019-21_total"])}\'\'\' ' + \
                  f'|| style="text-align:center;background:#eaf3ff;"|{if_value(ser["2019-21_male"])} ' + \
                  f'|| style="text-align:center;background:#fee7f6;"|{if_value(ser["2019-21_female"])} ' + \
                  f'|| style="padding-right:2ex;"|{if_value(ser["2019-21_fΔm"])}'

    if lang == 'ru':
        st = re.sub('(?<=\\d)\\.(?=\\d)', ',', st)  # replace . to comma, if this . is between two digits
        # st = st.replace('padding-right:1,5ex;', 'padding-right:1.5ex;')

    st = table_header + st + '\n|}'
    
    # gray color for missing values
    st = st.replace(';"|—', ';color:silver;"|—') \
           .replace(';"|\'\'\'—\'\'\'', ';color:silver;"|\'\'\'—\'\'\'')

    return st


table_code = create_table(df_NFHS, file_header='NFHS_header_ru -2019-21.txt', lang='ru')

# write the code to file
with open('output/Table code for Indian states by NFHS -ru.txt', 'w', encoding="utf-8") as fh:
    fh.write(table_code)
```

```python
def filter_df(df, selected_column):
    filtered_df = df.loc[:, [selected_column]]   \
                    .sort_values(by=selected_column, ascending=False) \
                    .dropna()

    filtered_df['group_label'] = filtered_df[selected_column].map(lambda x: f"{int(x):.1f}–{int(x) + 0.9:.1f}")
    
    
    # filtered_df['group_label'] = filtered_df['group_label'].replace(['79.5–79.99', '79.0–79.49', '78.5–78.99', '78.0–78.49'], '78.0–79.99')
    
    min_value = filtered_df[selected_column].min()
    max_value = filtered_df[selected_column].max()
    
    print(f"           ——— {selected_column} ———")
    print(f"Range: {min_value:.1f} – {max_value:.1f}   " +
          f"({filtered_df[selected_column].idxmin()} – {filtered_df[selected_column].idxmax()})")
    print(f"Number of groups: {filtered_df['group_label'].nunique()}")
    print(f"Number of values: {len(filtered_df)}")

    return filtered_df


df_sel = filter_df(df_map, selected_column='2019-21_total')
df_sel
```

```python
def extract_indexes(subdf, dd_legend = dd_legend):
    group_label = subdf['group_label'].iloc[0]
    countries = subdf.index.to_list()
    color = (dd_legend[group_label])
    
    ls_grouping.append(CountryGroup(group_label=group_label, countries=countries, color=color))

    return pd.Series([color, countries], index=['color', 'regions'])


ls_grouping = []
df_grouped = df_sel.groupby(['group_label'])[['group_label']].apply(extract_indexes).loc[::-1]

df_grouped
```
