```python
import pandas as pd
import math
import re
import itertools
from collections import namedtuple

import sys
sys.path.append("..")
import mal_moduls_private.mal_total as mal
```

```python
df = pd.read_excel(f'data/Life expectancy, Statistical Areas Level 4 - 2016-2018 to 2022-2024.xlsx', sheet_name='Table 1',
                   skiprows=8, skipfooter=1, usecols="B:AC", index_col=0)

df.index.name = ''

# drop empty records and columns
df.dropna(axis='index',   how='all', inplace=True)
df.dropna(axis='columns', how='all', inplace=True)

# rename columns
df.columns = [f'{period}_{sex}' for period, sex in itertools.product(('2016-18', '2017-19', '2018-20', '2019-21', '2020-22', '2021-23', '2022-24'), ('m', 'f', 't'))]

# change order of columns
df = df[[f'{period}_{sex}' for period, sex in itertools.product(('2016-18', '2017-19', '2018-20', '2019-21', '2020-22', '2021-23', '2022-24'), ('t', 'm', 'f'))]]

# filter records
states = ('Total New South Wales', 'Total Victoria', 'Total Queensland', 'Total South Australia', 'Total Western Australia',
          'Total Tasmania', 'Total Northern Territory', 'Total Australian Capital Territory', 'Australia')
df = df.loc[states, :]

df.fillna('')
```

```python
# rename indexes
df.rename(index = {'Total New South Wales': 'NSW',
                   'Total Victoria': 'VIC',
                   'Total Queensland': 'QLD',
                   'Total South Australia': 'SA',
                   'Total Western Australia': 'WA',
                   'Total Tasmania': 'TAS',
                   'Total Northern Territory': 'NT',
                   'Total Australian Capital Territory': 'ACT',
                   'Australia': '_AU_'                   
                  }, inplace=True)

# sort values
df.sort_values(by=['2020-22_t', '2020-22_m', '2020-22_f'], ascending=False, inplace=True)
df = pd.concat([df.loc[['_AU_']], df.drop('_AU_')])

df
```

```python
# create code for placing info in Wikipedia
def create_table(df, file_header, lang='ru'):

    def if_value(x, prec=1):
        return '—' if math.isnan(x) else \
               f"{x:0.{prec}f}"  if x>=0 else \
               f"−{-x:0.{prec}f}"                #"{x:0.{prec}f}".format(x, prec)
    
    def chval(x, prec=1, *, add_par=''):  # change_value
        return f'style="padding-right:1.5ex;{add_par}"| —' if math.isnan(x) else \
               f'style="padding-right:1.5ex;color:darkgreen;{add_par}"| {x:0.{prec}f}' if x>0 else \
               f'style="padding-right:1.5ex;color:crimson;{add_par}"| −{-x:0.{prec}f}' if x<0 else \
               f'style="padding-right:1.5ex;color:darkgray;{add_par}"| {x:0.{prec}f}'
    
    def chval_bold(x, prec=1, *, add_par=''):  # change_value
        return ' \'\'\'—\'\'\'' if math.isnan(x) else \
               f'style="padding-right:1.5ex;color:darkgreen;{add_par}"| \'\'\'{x:0.{prec}f}\'\'\'' if x>0 else \
               f'style="padding-right:1.5ex;color:crimson;{add_par}"| \'\'\'−{-x:0.{prec}f}\'\'\'' if x<0 else \
               f'style="padding-right:1.5ex;color:darkgray;{add_par}"| \'\'\'{x:0.{prec}f}\'\'\''

    with open('design/' + file_header, mode='r', encoding="utf-8") as fh:
        table_header = fh.read()

    st = ''
    for i in range(len(df)):
        ser = df.iloc[i]    
        if ser.name == '_AU_':
             st += '\n' + '|-class=static-row-header\n' + \
                  f'| \'\'\'{dd_replacement[ser.name][lang][0]}\'\'\' ' + \
                  f'||style="background:#e0ffd8;"| \'\'\'{if_value(ser["2019-21_t"])}\'\'\' ' + \
                  f'||style="background:#eaf3ff;"| \'\'\'{if_value(ser["2019-21_m"])}\'\'\' ' + \
                  f'||style="background:#fee7f6;"| \'\'\'{if_value(ser["2019-21_f"])}\'\'\' ' + \
                  f'||style="background:#fff8dc;"| \'\'\'{if_value(ser["2019-21_fΔm"])}\'\'\' ' + \
                  f'||{chval_bold(ser["Δ1"], add_par="border-left-width:2px;")} ' + \
                  f'||style="background:#e0ffd8;border-left-width:2px;"| \'\'\'{if_value(ser["2020-22_t"])}\'\'\' ' + \
                  f'||style="background:#eaf3ff;"| \'\'\'{if_value(ser["2020-22_m"])}\'\'\' ' + \
                  f'||style="background:#fee7f6;"| \'\'\'{if_value(ser["2020-22_f"])}\'\'\' ' + \
                  f'||style="background:#fff8dc;"| \'\'\'{if_value(ser["2020-22_fΔm"])}\'\'\' ' + \
                  f'||{chval_bold(ser["Δ2"], add_par="border-left-width:2px;")} ' + \
                  f'||style="background:#e0ffd8;border-left-width:2px;"| \'\'\'{if_value(ser["2021-23_t"])}\'\'\' ' + \
                  f'||style="background:#eaf3ff;"| \'\'\'{if_value(ser["2021-23_m"])}\'\'\' ' + \
                  f'||style="background:#fee7f6;"| \'\'\'{if_value(ser["2021-23_f"])}\'\'\' ' + \
                  f'||style="background:#fff8dc;"| \'\'\'{if_value(ser["2021-23_fΔm"])}\'\'\' ' + \
                  f'||{chval_bold(ser["Δ3"], add_par="border-left-width:2px;")} ' + \
                  f'||style="background:#e0ffd8;border-left-width:2px;"| \'\'\'{if_value(ser["2022-24_t"])}\'\'\' ' + \
                  f'||style="background:#eaf3ff;"| \'\'\'{if_value(ser["2022-24_m"])}\'\'\' ' + \
                  f'||style="background:#fee7f6;"| \'\'\'{if_value(ser["2022-24_f"])}\'\'\' ' + \
                  f'||style="background:#fff8dc;"| \'\'\'{if_value(ser["2022-24_fΔm"])}\'\'\' ' + \
                  f'||{chval_bold(ser["Δ_total"], add_par="border-left-width:2px;")}'
        else:
            name_link = dd_replacement[ser.name][lang][1]
            name_visible = dd_replacement[ser.name][lang][0]
            name_inserted = name_link if name_link == name_visible else f"{name_link}|{name_visible}"
            st += '\n' + '|-\n' + \
                  f'| [[{name_inserted}]] ' + \
                  f'||style="background:#e0ffd8;"| \'\'\'{if_value(ser["2019-21_t"])}\'\'\' ' + \
                  f'||style="background:#eaf3ff;"| {if_value(ser["2019-21_m"])} ' + \
                  f'||style="background:#fee7f6;"| {if_value(ser["2019-21_f"])} ' + \
                  f'||style="background:#fff8dc;"| {if_value(ser["2019-21_fΔm"])} ' + \
                  f'||{chval(ser["Δ1"], add_par="border-left-width:2px;")} ' + \
                  f'||style="background:#e0ffd8;border-left-width:2px;"| \'\'\'{if_value(ser["2020-22_t"])}\'\'\' ' + \
                  f'||style="background:#eaf3ff;"| {if_value(ser["2020-22_m"])} ' + \
                  f'||style="background:#fee7f6;"| {if_value(ser["2020-22_f"])} ' + \
                  f'||style="background:#fff8dc;"| {if_value(ser["2020-22_fΔm"])} ' + \
                  f'||{chval(ser["Δ2"], add_par="border-left-width:2px;")} ' + \
                  f'||style="background:#e0ffd8;border-left-width:2px;"| \'\'\'{if_value(ser["2021-23_t"])}\'\'\' ' + \
                  f'||style="background:#eaf3ff;"| {if_value(ser["2021-23_m"])} ' + \
                  f'||style="background:#fee7f6;"| {if_value(ser["2021-23_f"])} ' + \
                  f'||style="background:#fff8dc;"| {if_value(ser["2021-23_fΔm"])} ' + \
                  f'||{chval(ser["Δ3"], add_par="border-left-width:2px;")} ' + \
                  f'||style="background:#e0ffd8;border-left-width:2px;"| \'\'\'{if_value(ser["2022-24_t"])}\'\'\' ' + \
                  f'||style="background:#eaf3ff;"| {if_value(ser["2022-24_m"])} ' + \
                  f'||style="background:#fee7f6;"| {if_value(ser["2022-24_f"])} ' + \
                  f'||style="background:#fff8dc;"| {if_value(ser["2022-24_fΔm"])} ' + \
                  f'||{chval(ser["Δ_total"], add_par="border-left-width:2px;")}'

    if lang == 'ru':
        st = re.sub('(?<=\\d)\\.(?=\\d)', ',', st)  # replace . to comma, if this . is between two digits
        st = st.replace('padding-right:1,5ex;', 'padding-right:1.5ex;')

    st = table_header + st + '\n|}'
    
    # gray color for missing values
    st = st.replace(';"| —', ';color:silver;"| —')

    return st


table_code = create_table(df, file_header='Australia_header_ru -2024.txt', lang='ru')

# write the code to file
with open('output/Table code for Australian states -ru.txt', 'w', encoding="utf-8") as fh:
    fh.write(table_code)
```

```python
def filter_df(df, selected_column):
    filtered_df = df.loc[:, [selected_column]]   \
                    .sort_values(by=selected_column, ascending=False) \
                    .dropna()

    filtered_df['group_label'] = filtered_df[selected_column].map(lambda x: f"{x * 4 // 2 / 2:.1f}–{x * 4 // 2 / 2 + 0.4:.1f}")
    
    
    filtered_df['group_label'] = filtered_df['group_label'].replace(['79.5–79.9', '79.0–79.4', '78.5–78.9', '78.0–78.4', '77.5–77.9'], '77.5–79.9')
    
    min_value = filtered_df[selected_column].min()
    max_value = filtered_df[selected_column].max()
    
    print(f"           ——— {selected_column} ———")
    print(f"Range: {min_value:.1f} – {max_value:.1f}   " +
          f"({filtered_df[selected_column].idxmin()} – {filtered_df[selected_column].idxmax()})")
    print(f"Number of groups: {filtered_df['group_label'].nunique()}")
    print(f"Number of values: {len(filtered_df)}")

    return filtered_df


df_sel = filter_df(df_map, selected_column=SELECTED_COLUMN)
df_sel
```

```python
def extract_indexes(subdf, dd_legend = dd_legend):
    group_label = subdf['group_label'].iloc[0]
    countries = subdf.index.to_list()
    color = (dd_legend[group_label])
    
    ls_grouping.append(CountryGroup(group_label=group_label, countries=countries, color=color))

    return pd.Series([color, countries], index=['color', 'regions'])


ls_grouping = []
df_grouped = df_sel.groupby(['group_label'])[['group_label']].apply(extract_indexes).loc[::-1] \
                   .rename(index={'77.5–79.9': '<80'})

if ls_grouping[0].group_label == '77.5–79.9':
    ls_grouping[0] = ls_grouping[0]._replace(group_label = '<80')

df_grouped
```
